````python
#calculate the monthly expenses
#Scenario : you want to calculate your monthly expenses for rent, utilities,groceries, and entertenment
def calculate_expenses(rent, utilities, groceries, entertainment):
    return rent + utilities + groceries + entertainment
rent = int(input("Enter your rent: "))
utilities = int(input("Enter your utilities: "))
groceries = int(input("Enter your groceries: "))
entertainment = int(input("Enter your entertainment: "))
monthly_expenses = calculate_expenses(rent, utilities, groceries, entertainment)
print(f"Your monthly expenses are: ${monthly_expenses}")
````

````python
#calculate Temperature Units
#Scenario : You are building a weather app that takes temperature input in Celcius and convert it to Farenheit
def convert_to_fahrenheit(celsius):
    return (celsius * 9/5) + 32
celsius = float(input("Enter temperature in Celsius: "))
fahrenheit = convert_to_fahrenheit(celsius)
print(f"{celsius} degrees Celsius is equal to {fahrenheit} degrees Fahrenheit")
````

````python
#Discount Calculator for Shopping
#Scenario : You want to apply a discount to the total shopping bill based on a discount percentage
def apply_discount(total_bill, discount_percentage):
    discount_amount = total_bill * (discount_percentage / 100)
    return total_bill - discount_amount
total_bill = float(input("Enter the total bill amount: "))
discount_percentage = float(input("Enter the discount percentage: "))
discounted_bill = apply_discount(total_bill, discount_percentage)
discount_amount = total_bill - discounted_bill
print(f"You got ${discount_amount} as discount")
print(f"Total bill after discount ${discounted_bill}")
````

````python
#Distance Calculation between Two Points
#Scenario : In delivery services, we want to calculate the distance between two points to estimate delivery time
import math
def calculate_distance(X1,Y1,X2,Y2):
  return math.sqrt((X2-X1)**2 + (Y2-Y1)**2)
X1 = float(input("Enter the X-coordinate of the first point: "))
Y1 = float(input("Enter the Y-coordinate of the first point: "))
X2 = float(input("Enter the X-coordinate of the second point: "))
Y2 = float(input("Enter the Y-coordinate of the second point: "))
distance = calculate_distance(X1,Y1,X2,Y2)
print(f"distance between pickup point and Destination point is {distance}")
````

````python
#this will not work in google collab
import pyttsx3

def speak(text):
    engine = pyttsx3.init()
    engine.say(text)
    engine.runAndWait()

def generate_voice_invoice(item_name, quantity, price, tax_rate):
    subtotal = quantity * price
    tax_amount = subtotal * (tax_rate / 100)
    total = subtotal + tax_amount

    invoice_text = (
        f"Invoice generated. You ordered {quantity} {item_name}. "
        f"Price per item is {price} rupees. "
        f"Subtotal is {subtotal} rupees. "
        f"Tax amount at {tax_rate} percent is {tax_amount} rupees. "
        f"Your total payable amount is {total} rupees."
    )

    print(invoice_text)
    speak(invoice_text)

# Input
item = input("Item name: ")
qty = int(input("Quantity: "))
price = float(input("Price per unit: "))
tax = float(input("Tax rate %: "))

generate_voice_invoice(item, qty, price, tax)
````

````python
from gtts import gTTS
from IPython.display import Audio

def generate_voice(item_name, quantity, price, tax_rate):
    subtotal = quantity * price
    tax_amount = subtotal * (tax_rate / 100)
    total = subtotal + tax_amount

    print("\n---------Invoice----------")
    print(f"Item Name: {item_name}")
    print(f"Quantity: {quantity}")
    print(f"Price: ₹{price}")
    print(f"Subtotal: ₹{subtotal}")
    print(f"Tax {tax_rate}%: ₹{tax_amount}")
    print(f"Total: ₹{total}")
    print("--------------------------")

    # Create the voice message
    voice_message = (
      f"You ordered {quantity} {item_name}. "
      f"Each item costs {price} rupees. "
      f"So the total price without tax is {subtotal} rupees. "
      f"Including tax of {tax_amount} rupees, "
      f"your total payable amount is {total} rupees."
    )


    # Speak the invoice
    tts = gTTS(text=voice_message, lang='en')
    tts.save("order.mp3")
    return Audio("order.mp3", autoplay=True)


# --- Input Section ---
item_name = input("Enter the item name: ")
quantity = int(input("Enter the quantity: "))
price = float(input("Enter the price: "))
tax_rate = float(input("Enter the tax rate: "))

generate_voice(item_name, quantity, price, tax_rate)
````

````python
def pyramid(n):
  for i in range(1,n+1):
    print( " "* (n-i) + '*' * (2*i-1))
pyramid(5)
````

````python
def factorial(n):
  if n == 0:
    return 1
  else:
    fact = n * factorial(n-1)
    return fact
factorial(5)
````

````python
def fibonaci(n):
  if n == 0:
    return 0
  elif n ==1:
    return 1
  else:
    fibo = fibonaci(n-1) + fibonaci(n-2)
    return fibo
fibonaci(6)
````

````python
#Sum of a List of Number
def sum_list(numbers):
  if len(numbers) == 0:
    return 0
  else:
    return numbers[0] + sum_list(numbers[1:])
numbers = [1,2,3,4,5,6,7,8]
print(sum_list(numbers))
````
