1. Matplotlib:

The Foundation of Python Visualizations

What is it?

A low-level, highly customizable library for creating static, animated, and interactive plots

Key Features:
    
•Provides full control over every aspect Of a figure (axes, titles, labels, etc.).

•Supports a wide variety of plots (line, bar, scatter, pie, etc.), 

-Offers extensive customization options, though it may require more coding.

Use Case: Best when you need precise control or are creating custom visualizations from scratch.

2. Seaborn:

Built on Top of Matplotlib

What is it?

A high-level interface for drawing attractive and informative statistical graphics.
    
Key Features:
    
-Simplifies cornplex visualizations like heatmaps, pair and violin plots.
    
-Built.in themes and color palettes for aesthetically pleasing plots,
    
-Designed for working seamlessly with pandas DataFrarnes.
    
-Easily adds statistical summaries (e.g. confidence intervals).
    
Use Case:
    
Best for exploratory data analysis and when you want polished visuals with müünal code.

1️.Line Plot 📈

A line plot displays data points connected by straight lines.

It shows how one variable changes continuously with respect to another variable.

Why do we use a Line Plot?

-To observe trends (increase, decrease, fluctuations)

-To track changes over time

-To compare multiple trends on the same scale

When do we use a Line Plot?

-When the x-axis is continuous (time, distance, age, temperature)

-When data points have a natural order

-When you want to see patterns or movement

Example use cases

-Stock price over days

-Website traffic per hour

-Temperature changes during a day

When NOT to use

-For categorical (unordered) data

-For showing distribution of values

2️. Bar Plot 📊
What is a Bar Plot?

A bar plot uses rectangular bars to represent numerical values for categories.

Why do we use a Bar Plot?

-To compare quantities between different groups

-To easily see which category is largest or smallest

-To summarize data using aggregates (mean, sum, count)

When do we use a Bar Plot?

-When the x-axis is categorical

-When comparing groups or classes

-When values are discrete

Example use cases

-Sales by product category

-Number of students in each class

-Average salary by department

When NOT to use

-For continuous time-series trends

-When there are too many categories

3️. Scatter Plot 🔵
What is a Scatter Plot?

A scatter plot displays individual data points as dots on a 2D plane.

Why do we use a Scatter Plot?

-To understand relationship or correlation

-To detect patterns, clusters, or outliers

-To see if variables are positively or negatively related

When do we use a Scatter Plot?

-When both variables are numeric

-When you want to study correlation

-When exploring data before modeling

Example use cases

-Height vs weight

-Study hours vs exam score

-Price vs demand

When NOT to use

-When comparing categories

-When data is aggregated

```python
import matplotlib.pyplot as plt

x = [1,2,3,4]
y= [10,20,25,30]
#plot
plt.plot(x,y,color='blue',linestyle='--',marker='o',label='Growth')
plt.title("Matplotbil plot Example")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.legend()
plt.grid(True)
plt.show()
```

📉 Histogram
What is a Histogram?

A histogram is a plot that shows the distribution of a continuous numerical variable by dividing the data into bins (ranges) and displaying the frequency of values in each bin.

Why do we use a Histogram?

-We use a histogram to:

-Understand the distribution pattern of data

-See how values are spread

-Identify skewness (left-skewed / right-skewed)

-Detect outliers

-Understand central tendency and variability

When do we use a Histogram?

Use a histogram when:

-You have one continuous numeric variable

-You want to analyze frequency distribution

-You are performing Exploratory Data Analysis (EDA)

-Data is not categorical

Common Use Cases

-Exam score distribution

-Age distribution of people

-Salary or income distribution

-Temperature readings

When NOT to use a Histogram

-When data is categorical

-When comparing multiple categories (use bar plot)

-When you want exact values (histogram shows ranges, not exact counts)

```python
data=np.random.randn(1000)
plt.hist(data,bins=100,color="orange")
plt.title("Histogram - Matplotlib")
plt.xlabel("values")
plt.ylabel("Frequency")
plt.show()
```

```python
data=pd.DataFrame({'values':np.random.randn(1000)})
sns.histplot(data['values'],bins=30,kde=True,color="teal")
plt.title("Histogram -Seaborn")
plt.show()
```

🔥 Heat Map
What is a Heat Map?

A heat map is a graphical representation of data where values are shown using colors.
Each cell’s color indicates the magnitude or intensity of the value.

Why do we use a Heat Map?

-We use a heat map to:

-Quickly identify patterns and trends

-Easily compare many values at once

-Detect high and low intensity areas

-Visualize relationships in matrix-like data

When do we use a Heat Map?

-Use a heat map when:

-Data is in matrix or table format

-You want to visualize correlation

-You have large datasets that are hard to read as numbers

-You want color-based comparison

Common Use Cases

-Correlation matrix in data analysis

-Confusion matrix in machine learning

-Website user activity by time/day

-Sales intensity by region and product

When NOT to use a Heat Map

- When exact values are required
- When dataset is very small
- When color perception may mislead

Key Advantages

-Fast pattern recognition

-Handles large data well

-Visually intuitive

Key Limitation

-Colors may be misinterpreted

-Not ideal for precise comparisons

```python
import numpy as np
data = np.random.rand(5,5)
plt.imshow(data,cmap="coolwarm",interpolation='nearest')
plt.colorbar()
plt.title("Heatmap - Matplotlib")
plt.show()
```

```python
data = pd.DataFrame(np.random.rand(5,5),columns=list('ABCDE'))
sns.heatmap(data,annot=True,cmap='coolwarm')
plt.title("Heatmap - Seaborn")
plt.show()
```

📦 Box Plot (Box-and-Whisker Plot)
What is a Box Plot?

A box plot is a statistical chart that summarizes a dataset using five key values:

Minimum

-First Quartile (Q1)

-Median (Q2)

-Third Quartile (Q3)

Maximum

-It also highlights outliers.

Why do we use a Box Plot?

-We use a box plot to:

-Understand data spread and variability

-Identify outliers

-Compare distributions across categories

-See skewness of the data

When do we use a Box Plot?

-Use a box plot when:

-You want to summarize numerical data

-You need to compare distributions between groups

-You want to detect outliers

-Data is continuous

Common Use Cases

-Salary distribution by department

-Exam score comparison across classes

-Product price analysis

-Delivery time comparison

When NOT to use a Box Plot

- When exact values are required
- 
- When dataset is very small
- 
- When distribution shape details are important (use violin plot)

What Each Part Means

Box → Middle 50% of data (Q1 to Q3)

Line inside box → Median

Whiskers → Data range (excluding outliers)

Dots outside whiskers → Outliers

Key Advantage

-Compact summary of data

-Excellent for comparisons

```python
data = [np.random.rand(50) for _ in range(4)]
plt.boxplot(data,labels=['A','B','C','D'])
plt.title("Box Plot - Matplotlib")
plt.show()
```

```python
import numpy as np
data = pd.DataFrame({'Category':np.repeat(['A','B','C','D'],50),
                     'value':np.random.rand(200)
})
sns.boxplot(x="Category",y="value",data=data,palette='Set3')
plt.title( "Box plot - Seaborn")
plt.show()
```

```python
#pair plot

iris=sns.load_dataset('iris')
sns.pairplot(iris,hue='species',palette='husl')
plt.title("Pair plot - Seaborn")
plt.show()
```
