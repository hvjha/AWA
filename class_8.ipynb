```python
current_balance = float(345678.90)
acc_choice = input("what is your account type (saving/current): ").strip().lower()
if acc_choice == 'saving':
    daily_limit = 50000
elif acc_choice == 'current':
    daily_limit = 100000
else:
    print("Invalid account type ! please enter saving or current.")
exit()
print(f"Your per day withdraw limit is {daily_limit}")

print("Enter 'check' to check balance OR 'withdraw' to withdraw the balance OR 'deposite' to add money")
user_choice = input("Want to check_balance or withdraw_balance or add money: ").strip().lower()
if user_choice == 'check':
    print(f"Your current balance is: {current_balance}")
elif user_choice == 'withdraw':
    try:
        amount = float(input("Enter the amount: "))
        if amount <= 0:
            print("Invalid amount! Amount must be greater than 0.")
        elif amount > daily_limit:
            print(f"Daily limit exceeded! your per day withdraw limit is {daily_limit}")
        elif amount > current_balance:
            print("Insufficient balance!")
        else:
            current_balance -= amount
            print(f"Withdrawal successful! Remaining balance is {current_balance}")
    except ValueError:
        print("Invalid input! Please enter a numeric value.")
elif user_choice == 'deposite':
  try:
    amount = float(input("Enter the amount to deposite: "))
    if amount <= 0:
        print("Invalid amount! Amount must be greater than 0.")
    else:
        current_balance += amount
        print(f"Deposite successful! Remaining balance is {current_balance}")
  except ValueError:
        print("Invalid input! Please enter a numeric value.")

else:
    print("Invalid option! Please type 'check', 'withdraw' or 'deposite'")
```
