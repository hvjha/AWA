````python
num = int(input("Enter a number: "))
fact = 1
n = num
while num >0:
  fact = fact * num
  num = num - 1
print(f"factorial of {n} is {fact}")
````

````python
#reversed a number
number = int(input("Enter a number: "))
if number < 0:
  number = - number
reversed_number = 0
while number > 0:
  digit = number % 10
  reversed_number = reversed_number * 10 + digit
  number = number // 10
print(f"Reversed_number: {reversed_number}")
````
