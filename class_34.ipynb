Shallow Copy And Deep Copy

Shallow Copy

A shallow copy creates a new outer object, but does not create copies of nested objects.
    
Instead, it copies references to the inner objects.

Ways to create a shallow copy

copy() method (for lists, dicts, sets)

copy.copy()

List slicing [:]

Deep Copy

A deep copy creates a new outer object and new copies of all nested objects, recursively.

Way to create a deep copy

copy.deepcopy()

```python
#shallow copy

import copy

original = [[1, 2], [3, 4]]

shallow = copy.copy(original)

shallow[0][0] = 100

print("Original:", original)
print("Shallow :", shallow)
```

```python
original = [[1, 2], [3, 4]]

shallow = copy.copy(original)

shallow[0] = [1,3,6]
print("Original:", original)
print("Shallow :", shallow)
```

```python
original[0][0]=10
```

```python
print("Original:", original)
print("Shallow :", shallow)
```

key points:
The outer structure (list) is copied
Nested objects (inner list) are referenced, not copied

changes to nested objects in the shallow copy affect the original object.

```python
#deep copy
import copy

original = [[1, 2], [3, 4]]

deep = copy.deepcopy(original)

deep[0][0] = 100

print("Original:", original)
print("Deep    :", deep)
```

```python
id(original)
id(deep)
```

```python
id(original[0])
```

```python
id(deep[0])
```

```python
original = [[1, 2], [3, 4]]

deep = copy.deepcopy(original)

deep[0] = [10,67,89]

print("Original:", original)
print("Deep    :", deep)
```
