```python
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt
```

```python
np.random.seed(42)
battery_life = np.random.normal(loc = 9.5, scale = 1, size = 50)
battery_life[:10]
```

```python
plt.hist(battery_life,bins=10, color="skyblue")
plt.title("sample Battery Life")
plt.xlabel("Battery Life(Hours)")
plt.ylabel("Frequency")
plt.show()
```

```python
test_result = stats.ttest_1samp(battery_life,popmean=10)
test_result
```
