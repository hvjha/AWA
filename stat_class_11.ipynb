````python
import numpy as np
import matplotlib.pyplot as plt
#Number of dice per trail
dice_per_roll = 10
#Number of r trail
num_trails = 10000
#Roll 10 dice per trail, 10,000 times
#Each row is a trail of 10 dice
rolls = np.random.randint(1,7,(num_trails,dice_per_roll))
#compute the average of each trail
averages = rolls.mean(axis=1)
#plot the distribution of the averages
plt.hist(averages,bins=30,edgecolor="black", density=True)
plt.title(f"Distribution of Average of {dice_per_roll} Dice Rolls")
plt.xlabel("Average Roll Value")
plt.ylabel("Probability Density")
plt.show()
````

```
when individual events are messy and unpredictable,
their average behaves predictably,
forming the bell-shaped curve of the normal distribution
```
