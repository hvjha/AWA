```python
#Normal way
original_dict = {'a':1,'b':2,'c':3}
swap_dict = {}
for i,j in original_dict.items():
    swap_dict[j] = i
print(swap_dict)
```

```python
#using dictionary comprehension
original_dict = {'a':1,'b':2,'c':3}
swap_dict={i:j for j, i in original_dict.items()}
print(swap_dict)
```

Generating Reports

Aggregating and processing data for reporting

```python
sales = [
    {"category":"electronics","amount":1200},
    {"category":"clothing","amount":300},
    {"category":"electronics","amount":1800},
]
sales_summary = {item["category"]:sum(s["amount"] for s in sales if s["category"] ==item["category"]) for item in sales}
print(sales_summary)
```

```python
sales = [
    {"category":"electronics","amount":1200},
    {"category":"clothing","amount":300},
    {"category":"electronics","amount":1800},
]
sales_summary = {}
for sale in sales:
    category = sale["category"]
    amount = sale["amount"]
    if category in sales_summary:
        sales_summary[category]+=amount
    else:
        sales_summary[category] = amount
print(sales_summary)
```

Configuration Files and Metadata Extraction

```python
students = [
{"name": "Alice", "score":85},
{"name": "Bob", "score": 92},
{"name": "Charlie","score":78},
{"name": "David", "score": 95}
]
# Get names of students with scores above 98
top_students =[student["name"] for student in students if student["score"] > 90]
print (top_students)
```
