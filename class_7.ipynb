```python
#Modern and preffered way to format strings using f-strings
name = "Muskan Singh"
prof = "Business development Executive "
place = "IndiaMart"
Time = 12.9876
print(f"I am {name} and i worked as {prof} at {place} from last {Time:.2f} months")
```

```python
#Value Error
try:
  num = int(name)
  print(num)
except ValueError:
  print("Error: Invalid input, can't convert 'abc' to an integer")
```

```python
num1 = 10
num2 = 0
result = num1 / num2
print(result)
```

```python
#Zero division Error
try:
  num1 = 10
  num2 = 0
  result = num1 / num2
  print(result)
except ZeroDivisionError:
  print("Error: We can't divide any number with zero")
```
