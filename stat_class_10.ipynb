````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
````

````python
#Generate mostly normal scores
np.random.seed(42)
scores = np.random.normal(70,10,30)

#add outlier
scores = np.append(scores,120)

#put into a Data Frame
df = pd.DataFrame({'scores':scores})
df.head(5)
````

````python
plt.figure(figsize=(6,4))
sns.boxplot(y='scores',data=df,color='skyblue')
plt.title("Boxplot of student scores")
plt.show()
````

````python
import numpy as np
import pandas as pd

np.random.seed(42)

study_hours = np.random.normal(5, 1, 30)
scores = study_hours * 10 + np.random.normal(0, 5, 30)

# Add an outlier (studied 4 hours but got 120 score)
study_hours = np.append(study_hours, 4)
scores = np.append(scores, 120)

df2 = pd.DataFrame({'study_hours': study_hours, 'scores': scores})
print(df2)
````
