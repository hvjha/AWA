```python
import zlib

data = b"Hello world! Hello world! Hello world!"
compressed = zlib.compress(data)

print(compressed)
print("Original size:", len(data))
print("Compressed size:", len(compressed))
```

```python
decompressed = zlib.decompress(compressed)
print(decompressed)
```
