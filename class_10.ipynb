```python
#union operation
set_a= {"Aman","Harsh","Ankit"}
set_b= {"Muskan","priti","siya"}
set_c = set_a | set_b
print(set_c)
```

```python
#intersection operation
num_1 = {1,2,3,4}
num_2 = {3,5,6,7}
result = num_1 & num_2
print(result)
```

```python
#differences
ans = num_2 - num_1
print(ans)
```

```python
#symmetry
symmetry = num_1 ^ num_2
print(symmetry)
```
