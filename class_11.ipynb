```python
dict = {"name":"Harsh","id":1,"contact":987}
print(dict)
print(dict.keys())
print(dict.values())
print(dict.items())
#accessing with key
print(dict["name"])
#accessing with get method
print(dict.get("id"))
#add a new key:value pair
dict["email"]="ajha@gmail.com"
dict["email"] = "jhahv@gmail.com"
print(dict)
#pop -> Remove the item with key
#popitem -> Remove the last inserted key value pair
#del -> del that perticular item whose key we provide
#clear -> clear the dictionary

dict.pop("id")
print(dict)
dict.popitem()
print(dict)
del dict["name"]
print(dict)
dict.clear()
print(dict)
```

```python
dict1 = dict(name="Harsh",id = 123,contact = 987)
print(dict1)
print(dict1.keys())
print(dict1.values())
print(dict1.items())
print(dict1["name"])
print(dict1.get("id"))
dict1["email"]="ajha@gmail.com"
dict1["email"] = "harsh@gmail.com"
print(dict1)
dict1.popitem()
print(dict1)
del dict1["name"]
print(dict1)
dict1.clear()
print(dict1)
```
