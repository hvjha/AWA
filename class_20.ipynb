```python
datetime.date:Represent a date (year,month,day)
datetime.time:Represent a time(hour,minute,second,microsecond)
datetime.datetime:Represents both date and time
datetime.timedelta:Represent a duration(difference beteen two date/time)
datetime.tzinfo:used for timezone information
```

```python
now = datetime.now()
```

```python
print(now)
```

```python
utc_now = datetime.utcnow()
print(utc_now)
```

```python
formated_date = now.strftime("%Y-%m-%d %H:%M:%S")
print(formated_date)
```

```python
parsed_Date = datetime.strptime("2025-11-25 15:49:43","%Y-%m-%d %H:%M:%S")
print(parsed_Date)
```

```python
new_format = parsed_Date.strftime("%d-%m-%Y")
print(new_format)
```

```python
from datetime import timedelta
```

```python
future_date = today + timedelta(days = 10)
print(future_date)
```

```python
time_difference = future_date - today
print("day difference: ",time_difference.days)
```
