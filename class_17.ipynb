```
Decorator
decorator do something similar - they let us add extra behavior to function without changing the function itself
```

````python
def add_greeting(func):
  def wrapper(name):
    print("Hello",end =" ")
    func(name) #original function
  return wrapper
````

````python
#Now ,we'll use it to "decorate" different functions:
@add_greeting
def welcome_msg(name):
  print(f"{name}, welcome to the event!")

@add_greeting
def goodbye_msg(name):
  print(f"{name}, have a great day!")
````

**How func() Works**

func as a parameter: when we apply @add_greeting to a function, that function (e.g., welcome_msg) is passed as an argument to add_greeting and becomes func inside it..

calling func() Inside wrapper: Inside the wrapper, we use func(name) to call the original function(welcome_msg) with the provided argument(name).

This line allows the original function to run with its normal behavior while being wrapped by additional code(in this case, print("Hello",end =" ")).

**Real-Life Analogy**

Security check at an event
Imagine you're hosting an exclusive event where only guests with an invitation can enter.
You have a list of activities (functions) that are accessible only to invited guests.
Instead of adding a security check at every activity, you can create a decorator that acts like a security guard.
This decorator Will Check if a guest has an invitation before they can enter any activity.

````python
#we'll create a check_invitation decorator that verifies if a person has an invitation before letting them access certain function
#The decorator function

def check_invitation(func):
  def wrapper(guest):
    if guest.get("invited"):
      return func(guest)
    else:
      print(f"soory {guest['name']}, you are not invited to this event")
  return wrapper

#Decorator function
@check_invitation
def entry(guest):
  print(f"Welcome, {guest['name']} !Enjoy the event")

@check_invitation
def access_vip_area(guest):
  print(f"{guest['name']}, you have access to the VIP area")

#Guests
guest1 = {"name":"Harsh", "invited":True}
guest2 = {"name":"Muskan", "invited":False}

#calling  function
entry(guest1)
entry(guest2)

access_vip_area(guest1)
access_vip_area(guest2)
````
