```
When is each measure most useful ?
  Mean - data is nice and balanced
  Median - there are outlier or data is skewed
  Mode - when we have categorical data or repeated items
```

```
  Skewed?
    data isn't balanced - its pulled more to one side that's the called skewed data
  Right skewed ?
    the tail is on the right side
    most value are small, but a few are much larger
    income
  Left skewed ?
    the tail is on the left side
    most value are high, but a few are very low
    retirement age
```

```
Mean ?
  Best for : symetrical data
  weakness : distorted by extremes

Median ?
  Best for : skewed data
  weakness : ignores how far value are apart

Mode ?
  Best for : Categorical, repeated data
  weakness: might not exist, might be multiple
  shoe size
```

````python
#Disperssion
import numpy as np
classA = [70,71,69,69,71]
classB = [40,90,50,80,60]

print("Mean A: ",np.mean(classA))
print("Mean B: ",np.mean(classB))
````

```
class A's score are all near 70 - prety Consistent
Class B's score bounces all over the place from 40 to 90

conclusion:
  boyh class perform around the same level, class B is way more inconsistent

Disperssion helps answer ?
  how consistent or varied is the data
```

```
  the easiest way to measure disperssion - the range
  simply maximum - minimum

  rangeA = max(classA) - min(classA)
  rangeB = max(classA) - min(classB)
```

````python
rangeA = max(classA) - min(classA)
rangeB = max(classB) - min(classB)
rangeA,rangeB
````

```
class B result is more spread out - more variability, less consistent

The smaller the range , the more consistent the data

The range only looks at two values - heighest and the lowest what about all the score in between?

then standard Deviation, variance comes into picture
```

````python
classA = [70,71,69,70,71]
classB = [40,90,50,80,60]

np.var(classA), np.var(classB)
````

```
  the variance for classA is 0.56 but for classB its 344 - huge difference,

  Now let's take the square root to get the standard deviation

```

````python
np.std(classA),np.std(classB)
````

```
Class A's SD = 0.75
Class B's SD = 18.54

  That means
    class A score are all close to the mean (very stable)

    class B score are far from the mean (more spread out)

    higher SD means more variability - less consistency

    Teacher, low SD , Student perform similarly
    Investor, high SD, Means stock price unpredictable(risky)

    manufacture, low SD, Means products are uniform in quality
  ```
