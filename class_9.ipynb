```python
#Checking for palindrome
str = input("Enter a string: ")
str1 = str[::-1]
if str == str1:
    print("Palindrome")
else:
  print("Not Palindrome")
```

```python
user_input = input("Enter a string or number: ")
try:
    if user_input.isdigit():
        user_input = int(user_input)
        original = str(user_input)
        reverse = original[::-1]

        if original == reverse:
            print("Palindrome Number")
        else:
            print("Not a Palindrome Number")
    elif user_input.isalpha():
        original = user_input.lower()
        reverse = original[::-1]

        if original == reverse:
            print("Palindrome String")
        else:
            print("Not a Palindrome String")
    else:
        raise ValueError("Input must contain only letters OR only digits.")

except ValueError:
    print("Invalid input! Please enter either string or Number")
```
