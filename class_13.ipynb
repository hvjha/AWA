````python
#program to find the common element between a list and a tuple using loop
l = [1,2,3,4,5,6,7]
t = (1,2,7,8,9,11,14)
for i in l:
  for j in t:
    if i == j:
      print (f"common elements between list and tuples {i}")
````

````python
l = [1,2,3,4,5,6,7]
t = (1,2,7,8,9,11,14)

common_ele = []

for i in l:
  if i in t:
    common_ele.append(i)
print(common_ele)
````

````python
l = [1,2,3,4,5,'p',['a',4,5,(1,"c","d",6)]]
l1= []
for i in l:
  if type(i) == str:
    l1.append(i)
  elif type(i) == list:
    for j in i:
      if type(j) == str:
        l1.append(j)
      elif type(j) == tuple:
        for k in j:
          if type(k) == str:
            l1.append(k)
print(l1)
````
