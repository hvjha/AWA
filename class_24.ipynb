series means one dimension labeled array, capable of holding any data type it can be integer

````python
s = pd.Series([3,-5,7,4])
print(s)
````

```
loc - label based Indexing
iloc - integer(Position) based indexing

In loc when we perform slicing then end point is
Included but in iloc excluded the end point
```

````python
data = {'Name':['Aman','Harsh','Ankit','Adarsh'],'Age':[23,24,22,21], 'Id':[1,2,3,4]}
df = pd.DataFrame(data,index=['a','b','c','d'])
print(df)
````

````python
df.head()
````

````python
df.loc['a']
````

````python
df.iloc[1]
````

````python
df.loc['a':'c']
````

````python
df.iloc[0:2]
````
